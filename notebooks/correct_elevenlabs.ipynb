### 2. Upload the two files

```python
from google.colab import files
import glob, os
os.makedirs('work', exist_ok=True)
os.chdir('work')
files.upload()
print(glob.glob('*'))
os.chdir('..')
```

### 3. Open the correction page
Wait for the link to appear, then tap it. If it says *no sessions*, reload the page once.

```python
import glob, subprocess, time
from google.colab import output
video = [f for f in glob.glob('work/*') if not f.endswith('.json')][0]
transcript = glob.glob('work/*.json')[0]
print('video:', video, '| transcript:', transcript)
server = subprocess.Popen(['python', '-u', '-m', 'transcribe.app', '--no-browser', '--open-elevenlabs', '../' + transcript, '--audio', '../' + video], cwd='transcribe')
time.sleep(40)   # decoding an hour of audio takes about half a minute
output.serve_kernel_port_as_window(8000)
```

**On the page:** open the session, then correct chunk by chunk: **Esc** play/pause, **F1/F2** back/forward 3 s, **F3/F4** slower/faster, **Ctrl+Enter** save and next, **Alt+N** next flagged chunk. A word you fix twice is fixed everywhere automatically.

When finished: **Download TXT / SRT / JSON** (top right).
